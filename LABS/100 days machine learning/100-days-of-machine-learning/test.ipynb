```python
import pandas as py
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
print("hell0=o")
```

```python
df = py.read_csv('train.csv')
```

```python
print(df.head)
```

```python
print(df.info)
```

```python
# Check for missing values
print("Missing values per column:")
print(df.isnull().sum())

# Check for duplicate rows
print(f"Number of duplicate rows: {df.duplicated().sum()}")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Histogram for Age
sns.histplot(df['Age'], kde=True, color='skyblue')
plt.title('Age Distribution')
plt.show()

# Boxplot for Fare
sns.boxplot(x=df['Fare'])
plt.title('Fare Boxplot')
plt.show()

# KDE Plot for Parch
sns.kdeplot(df['Parch'], fill=True, color='green')
plt.title('Parch Distribution')
plt.show()
```

```python
# Countplot for Pclass
sns.countplot(x='Pclass', data=df, palette='viridis')
plt.title('Passenger Count by Class')
plt.show()

# Pie Chart for Survival
survival_counts = df['Survived'].value_counts()
labels = ['Did Not Survive', 'Survived']
plt.pie(survival_counts, labels=labels, autopct='%1.1f%%', colors=['lightcoral', 'skyblue'])
plt.title('Survival Proportion')
plt.show()
```

```python
#task 5
# Scatterplot: Age vs Fare
sns.scatterplot(x='Age', y='Fare', data=df)
plt.title('Age vs Fare')
plt.show()

# Countplot: Survival by Pclass
sns.countplot(x='Pclass', hue='Survived', data=df)
plt.title('Survival by Passenger Class')
plt.show()

# Boxplot: Age by Survival Status
sns.boxplot(x='Survived', y='Age', data=df)
plt.title('Age Distribution by Survival Status')
plt.show()
```

```python
#task6
# Pairplot for key columns
sns.pairplot(df[['Age', 'Fare', 'Pclass', 'Survived']], hue='Survived', palette='coolwarm')
plt.suptitle('Pairplot of Selected Features', y=1.02)
plt.show()

# Heatmap for correlations
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```
