# Task 1(Function of Factorial)

```python
def fact(n):
    if n < 0:
        return("Factorial is not defined for given number.")
    elif n == 0 or n == 1:
        return(1)
    else:
        fact = 1
        for i in range(1,n+1):
            fact *= i
        return fact
num = int(input("Enter a number to calculate its factorial: "))
print(f"Factorial of {num} (recursive): {fact(num)}")
```
