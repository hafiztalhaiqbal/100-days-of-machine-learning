# Examples of Basic File Handling

## Example no 1 (CSV File)

### Creating a CSV FIle 

```python
import csv 
# Data to write (list of lists) 
data = [ 
["Name", "Age", "City"],  # Header 
["John", 25, "New York"], 
["Anna", 30, "Los Angeles"], 
["Mike", 35, "Chicago"] 
]
# Creating and writing to a CSV file 
with open('General.csv', mode='w', newline='') as file:
    writer = csv.writer(file) 
    writer.writerows(data)  # Write multiple rows 
print("CSV file 'General.csv' created and data written successfully!") 
```

### Reading a CSV File

```python
import csv 
with open('General.csv',mode='r') as file:
    reader = csv.reader(file)
    for rows in reader:
        print(rows)
```

### Appending a CSV File 


```python
import csv 
new_data = ['Sara',25,'Chicago']

with open ('General.csv',mode='a',newline='') as file:
    writer = csv.writer (file)
    writer.writerow(new_data)
print("New Data is successfully added to file")
```

## Example no 2 ( Json File)


### Creating a Json File

```python
import json
# data to be written 
data = {
    "people":[
        {"name":"John","age":25,"city":"New York"},
        {"name": "Anna", "age": 30, "city": "Los Angeles"}, 
        {"name": "Mike", "age": 35, "city": "Chicago"} 
    ]
}

# creating and writing data to json
with open('people_data.json',mode='w') as file:
    json.dump(data,file,indent = 4)
print("File is created")


        
```

### Reading a Json File

```python
import json
with open('people_data.json',mode='r') as file:
    data = json.load(file)
print(data)
```

### Appending Json File 

```python
import json
new_entry = {"name":"Sarah","age":35, "city":"chicago"}
with open('people_data.json',mode='r') as file:
    data=json.load(file)
data["people"].append(new_entry)

with open('people_data.json',mode='w') as file:
    json.dump(data,file, indent=4)
print("Appending is done")
    
```
