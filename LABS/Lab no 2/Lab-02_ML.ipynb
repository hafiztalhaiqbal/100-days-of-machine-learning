```python
plt.figure(figsize=(8,6))
sns.scatterplot(data=df, x='Age', y='Fare', hue='Survived', alpha=0.6)
plt.title('Scatter Plot of Age vs Fare')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.legend(title="Survived")
plt.show()
```

```python
df[['Age', 'Fare']].corr()
```

```python
# Grouped bar chart
plt.figure(figsize=(8,6))
sns.countplot(data=df, x='Pclass', hue='Survived')
plt.title('Survival Count by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Count')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()

# Tabulating survival counts across classes
print(df.groupby(['Pclass', 'Survived']).size().unstack())
```

```python
# Boxplot for Age across survival status
plt.figure(figsize=(8,6))
sns.boxplot(data=df, x='Survived', y='Age', palette='Set2')
plt.title('Age Distribution Across Survival Status')
plt.xlabel('Survived')
plt.ylabel('Age')
plt.xticks([0, 1], ['No', 'Yes'])
plt.show()
```
